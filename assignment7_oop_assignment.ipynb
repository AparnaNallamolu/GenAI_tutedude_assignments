## Task 1 - Basic Class and Object Creation

```python
'''
Create a class product with:

    Attributes: name, price, category
    method: get_info() --> prints product details
    Create 2 objects and call get_info().

Extra (optional): Add a method apply_discount(percent) that returns the discounted price
'''
```

```python
class Product:

    def __init__(self, name, price, category, percent):
        self.name = name
        self.price = price
        self.category = category
        self.percent = percent

    def get_info(self):
        print(f'The product {self.name} is of price {self.price} and comes under category {self.category}')

    def apply_discount(self):
        integer_convert = self.percent
        integer_price = self.price
        discount_percent = integer_convert / 100
        discount = integer_price * discount_percent
        #print(discount_percent)
        #print(discount)
        print(f'The final price after giving {self.percent} % is {discount}')
```

### Task 3: Inheritance (Single Level)

```python
''' 
Create a subclass ElectronicProduct that inherits from Product.
Additional attribute:
Warranty_years
Override the get_info() method to include warranty info
Create an object and demonstrate inheritance + overriding
'''
```

```python
class ElectricProduct(Product):

    def __init__(self, name, price, category, percent, warranty_years):
        super().__init__(name, price, category, percent)
        self.warranty_years = warranty_years

    def get_info(self):
        super().get_info()
        print(f' Warranty: {self.warranty_years} months')
```

### Assigment 5 Abstraction (suing Abstract Base class)

```python
'''
Create an abstraction class  Payment with abstract method
process_payment(amount)

Then create two subclasses:
creditcardpayment
upi payment
both override process_payment() with simple print statements
test all classes
'''
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    def __init__(self, amount):
        self.amount = amount

    @abstractmethod
    def process_payment(self, amount):
        print(f'The process amount is {self.amount}')


class CreditCardPayment(Payment):

    def process_payment(self, amount):
        print(f'The process amount is {self.amount}')

class UPIPayment(Payment):

    def process_payment(self, amount):
        print(f'The process amount is {self.amount}')
```

```python
credit = CreditCardPayment(5000)
```

```python
credit.process_payment(5000)
```

```python
upi = UPIPayment(1500)
```

```python
upi.process_payment(1500)
```

### Task 6 Magic Methods and Operator Overloading

```python
'''
Add the following to your product class

        1. __str__ method
        return a readable string
        product(name, price, category)

        2. Operator overloading (__add__)
        Allow:
        product 1 + product2
        to return a the total combined price
        test this with two product objects

'''
```

```python
class Product2:

    def __init__(self, name, price, category):
        self.name = name
        self.price = price
        self. category = category

    def __add__(self, other):
        return Product2(self.name + " & " + other.name,  self.price + other.price, self.category + " & " + other.category)

    def __str__(self):
        return f'{self.name} combine price is {self.price} of category {self.category}'
```

```python
laptop = Product2('Laptop', 85000, 'Electronics')
Mobile = Product2('Mobile', 30000, 'Electronics')

```

### Task 7 Mini Project Simple Inventory System (OOPs Only)

```python
'''
Create 2 classes:

Class Inventory:

    Attributes:

        product --> list to store product objects

    Method:

        add_product(product)
        remove_product (name)
        get_total_value() --> sum prices of all the products
        show_all_products() --> print info for each product

class Store:

    Attribute:

        store_name
        inventory --> an inventory object

    Methods:

        add_new_product() ----> Takes input and creates product object
        show_summary()  --> prints total items and value

    Important: Use only OOPs concept -- no file handling no exception, no packages

    Test the system by:
    1. creating a store object
    2. adding 3 product
    3. show summary
    4. using __add__ to combine prices of two products

'''
```

```python
class Inventory:

    def __init__(self):
        self.product = []

    def add_product(self, product_obj):
      
        self.product.append(product_obj)

    def remove_product(self, name):
        for i in product_obj:
            if i.name == name:
                self.product.remove(i)
                break
        
    def get_total_value(self):
        total_item = 0
        for i in self.product:
            total_item = total_item + i.price
        return total_item

    def show_all_products(self):
        for i in self.product:
            print(f'Item name: {i.name} - price: {i.price}')

class Store():
    def __init__(self, name):
        self.store_name = name
        self.inventory = Inventory()

    def add_new_product(self):
        name = input(f"Enter the product name: ")
        price = float(input(f'Enter price for {name} product: '))

        class Product:

            def __init__(self, n, p):
                self.name = n
                self.price = p
            
            def __add__(self, other):
                return self.price + other.price

        new_item = Product(name, price)
        self.inventory.add_product(new_item)
        
    
    def show_summary(self):
        self.inventory.show_all_products()
        print(f'Total Value is: {self.inventory.get_total_value()}')


shop = Store('My shop')

for i in range(5):
    shop.add_new_product()

shop.show_summary()

product1 = shop.inventory.product[0]
product2 = shop.inventory.product[1]
addi = product1 + product2

print(f'the sum of {product1.name} and {product2.name} is or Rs. {addi}')



```
