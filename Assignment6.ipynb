### Task 1 Safe division Utility

```python
'''
Write a program that:
    1. Takes two inputs from the user numerator and denominator
    2. Use try - except to handle:
        1.  Value Error ( if input is not a number)
        2. Zero division error (if denominator - 0)
    3. if no error occurs print the result inside the else block
    4. in the finally block print "operation complete"
'''
```

```python
try:
    numerator = int(input('Enter the numerator value: '))
    denominator =  int(input('Enter the denominator value: '))

    result = numerator / denominator

except ValueError as ve:
    print(ve)

except ZeroDivisionError as zde:
    print(zde)

else:
    print(f"The result of {numerator} and {denominator} is ", result)
    

finally:
    print('operation compete')


    
```

### Task 2: Bill Calculation with error handling

```python
'''
Given a list of product prices:
prices = [120, 350,'abc',500,-200,800]
write code that:
    1. Iterates throught the list
    2. Tries to add only valid (positive numeric) prices to the total
    3. Handles:
        Type Error if value is not a number
        Custom exception using raise ValueError ('Negative price not allowed)
    4. prints the running total 

Excepted behavior:
    Skip invalid items but continue processing.
'''
```

```python
prices = [120, 350,'abc', 500, -200, 800]
result = 0
for p in prices:

    try:
        if p<0:
            raise ValueError ('Negative price not allowed')
            
       
        result+=p
        #print(result)



    except TypeError as te:
        print(te)  

    except ValueError as ve:
        print(ve)     

print(result)
```

### Task 3: Custom Exception : Age Validator

```python
'''
1Write a function check_age(age) that
    rasises a custome exception value error ('Age must be between 1 and 120) if age is out of range

2. In your main code:
    take age input form the user
    Use Try - except to catch and print the custom error message

'''
```

```python
def check_age(age):
    if age<1 or age>120:
        raise ValueError('Age must be between 1 and 120') 
    
    else:
        print(f'your age is {age}')

try:
    ag = int(input("Enter your valid Age: "))
    check_age(ag)

except ValueError as ve1:
    print(ve1)
```

### Task 4: File reader with exception handling


```python
'''
1. Ask the user for a filename
2. Try to open and read the file.
3. Handle:
    filenotfounderror
    PermissionError
4. If successful print first 3 lines of the file.
5. Use Finally to print "File operation attempted"
'''
```

```python
filename = input("Enter the file name: ")

try:
    with open(filename,'r') as rf:
        result1 = rf.readlines()[0:3] 
        print(result1)
       

except FileNotFoundError as fnfe:
    print(fnfe)

    
except PermissionError as pe:
    print(pe)

finally:
    print('File operation Attempted')
```

### Task 5: Mini Program: Safe Shoping Cart

```python
'''
Create a program that:
    1. has a cart list --> cart = []
    2. Runs a loop asking user to enter prices
    3. stops when users enter quit ('q')
    4. Instide the loop:
        Converti int to a float
        Handle value error if user intries invalid number
        raise a custom exception if price is negaitve
    5. at the end, print 
        Total Items
        Total Bill
'''
```

```python
cart = []

while True:
    val = input("Enter the price value:")
    
    if val == 'q':
        break

    try:
        prices = float(val)

        if prices < 0:
            raise ValueError('User enters invalid input')

        cart.append(prices)
        print('Items added: ', prices)

    except ValueError as vee:
        print(vee)

print('Total Items', len(cart))
print('Total bill', sum(cart))
    
```
