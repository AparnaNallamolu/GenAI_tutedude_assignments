### Task 6 percentiles and sorting

```python
'''
Using the same marks arrays:

1 . Sort the array
2. Find:

        25th percentile
        50th percentile
        75th percentile

3. Cont how many students scored above the average marks
'''
```

```python
marks
```

```python
np.sort(marks)
```

```python
np.percentile(marks, 25)
```

```python
np.percentile(marks, 50)
```

```python
np.percentile(marks, 75)
```

```python
count = 0
mean = np.mean(marks)
for i in marks:
    if i > mean:
        #print(i)
        count += 1
print(count)
```

```python
mean = np.mean(marks)
count = np.sum(marks > mean)
print(count)
```
