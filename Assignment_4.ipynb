### Task 7 Mini Project - Export discounted price

```python
'''
Create a dictionary:
prices = {"mouse":500,
          "Keyboard":800,
          "Monitor":7000,
          "Pendrive":400,
          "Camera":5000
          }

Ast the user for a discount percentage.
Write discounted price into discount_report.txt using
    product | Original Price | Discounted Price
After writing read the file and print to the terminal.
Extra (optional): Write a summary at the bottom of the file
Total Items: X
Average discounted Price: Y
'''
```

```python
import os

prices = {"mouse":500,
          "Keyboard":800,
          "Monitor":7000,
          "Pendrive":400,
          "Camera":5000
          }

discount_val = float(input("Enter the discount value: "))
discount = (100 - discount_val) / 100
print(discount)
```

```python
total_sum_discounted = 0
total_items = len(prices)

with open('discount_report.txt', 'w') as discount_calc:
    discount_calc.write(' Product    |   Original Price  |   Discount Price ')
    for product, org_price in prices.items():
        disc_price = org_price * discount
        total_sum_discounted += disc_price

        discount_calc.write(f'{product} |   {org_price} |   {disc_price}\n')
    
    avg_discount = total_sum_discounted / total_items
    discount_calc.write(f"Total Items: {total_items}\n")
    discount_calc.write(f"Average discount Price: {avg_discount}\n")
```
