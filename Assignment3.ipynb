### Task 1 basic function : Price after discount

```python
"""Write a function apply_discount(price,discount_percent) taht:
        1. Return the price after discount
        2. if discount_percer is missing, apply a default discount of 5%
        3. test the function with:
            apply_discount(1000, 10)
            apply_discount(500) #uses default discount
        Extra ( optional ): Add a condition inside the function to ensure discount never exceed 60%
"""
```

```python
def apply_discount(price, discount_percent=None):
    
    if discount_percent is None:
        discount_percent = 5
    
   
    if discount_percent >= 60:
        print('Notice: Discount capped at 60%')
        discount_percent = 60
    
    
    discount = price * (discount_percent / 100)
    final_price = price - discount
    
    return final_price


```

```python
apply_discount(1000, 10)
```

```python
apply_discount(500)
```

```python
apply_discount(500, 80)
```

### Task 2 Recursive Function: Factorial Utility

```python
"""
Create a recursive fucntion factorial(n) that:
    return the factorial n
    handle two edge case n==0 and n==1
    print an error message if n is negative.
test with:
    factorial(5)
    factorial(0)
    factorial(-3)
"""
```

```python
def factorial(n):
    if n==0 or n==1:
        number = 1
        print(f"The factorial value of {n} is: {number}")
    elif n<0:
        print(f'The number is negative')
    else:
        fact = 1
        for i in range(1,n+1):
            fact=fact*i
        print(f"the Factorial of num is: {fact}")
```

```python
factorial(5)
```

```python
factorial(0)
```

```python
factorial(-3)
```

### Task6 - Combined Utility Function

```python
'''
Write a function process_prices(prices) that:
1 taake a list of prices.
2. uses map() + lambda to apply 10% discount to all prices.
3. uses filter() to keep only discounted prices that are above 300.
4. return both lists:
discount prices
filtered prices
Test with:
process_prices([100,500,900,50,750])
'''
```

```python
def process_prices(prices):
    #Discount list
    discount_list = list(map(lambda x:x + (0.10 * x), prices))
    print("The discount list is: ", discount_list)

    #Filter prices above 300
    filtered_list = list(filter(lambda y: y>300, discount_list))
    print("The list above 300 is: ", filtered_list)
```

```python
process_prices([100,500,900,50,750])
```

### Task 7: Mini Project: Menu Using Functions

```python
'''
Create the following three small functions:
1. add_price(price_list, price) --> adds price to the list.
2. get_average_price(price_list): return average price
3. get_max_price(prices_list) --> return the maximum price
Then create a simple menu:
    1 -> add price
    2 -> show average price
    3 -> show highest price
    q -> quit
'''
```

```python
def adding(price_list, price):
    price_list.append(price)

def average(price_list):
    return sum(price_list) / len(price_list)

def max_price(price_list):
    return max(price_list)


prices = []
while True:
    print('1: Add price, 2: Show average price, 3: show highest price, q: Quit')

    choose = input('Enter your choice: ')

    if choose =='1':
        values = int(input("Enter the amount:"))
        adding(prices, values)
        print(f"{values} is added")

    elif choose == '2':
        avg = average(prices)
        print(f"Average is: {avg}")
    
    elif choose == '3':
        highest = max_price(prices)
        print(f'Highest price is: {highest}')
    
    elif choose == 'q':
        print('The program is ended...!')
        break

    else: 
        print('The option you entered is invalid')
```
