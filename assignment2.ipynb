### Task 2 Process multiple orders (for loop)

##### 1. given a list of order amounts: orders = [1200,2500,800,1750,3000], use a for loop to apply the discount rules from task1 to each order and print a summary table showing order_amount --> discount --> final amount.

##### 2. Also compute  and print the total revenue after discounts

##### extra (optional): print the number of orders tht received a discount (discount > 0).

```python

orders = [1200,2500,800,1750,3000]
total_revenue = 0
count=0
print('order_amount | discount | final amount')
for order_amount in orders:
    if order_amount >= 2000:
        discount=order_amount*15/100
        sub_total= order_amount - discount  
        #print(f'total amount paid: {order_amount} after discount of 15 % is: {sub_total}')
    elif order_amount>=1500 and order_amount < 2000:
        discount=order_amount*10/100 
        sub_total= order_amount - discount    
        #print(f'total amount paid: {order_amount} after discount of 10 % is: {sub_total}')
    elif order_amount>=1000 and order_amount < 1500:
        discount=order_amount*7/100
        sub_total= order_amount - discount   
        #print(f'total amount paid: {order_amount} after discount of 7 % is: {sub_total}')
    elif order_amount < 1000:
        discount=0
        sub_total=order_amount - discount 
        #print(f'total after paid {order_amount} with no discount is: {sub_total}')
    
    print(f'{order_amount} | {discount} | {sub_total}')
    total_revenue+=sub_total
    if discount>0:
        count+=1
print(f'The number of discounts you get: {count}')
print(f'The total revenue of all the orders after discount is: {total_revenue}')
    
    

```

### Task 3 User menu while loop + break /continue

```python
'''
1. Implement a simple while loop driven menu that keeps asking the user to choose 
an action until they press q to quit
Menu options:
    1. add order amount to a runnint list (use input())
    2. show all orders and total after applying discounts.
    q - quit

2. Use continue to reshow the menu after valid input and break to exit on q.
Constrain keep the logic simple and use only lists and loops ( no funcitons required)
'''
```

```python
orders = [] # This is our storage list
total_revenue = 0

while True:
    choice = input("Enter your choice 1 (Add), 2 (Summary), q (Quit): ")

    if choice == 'q':
        break
        
    elif choice == '1':
        amount = int(input('Enter the order amount: '))
        orders.append(amount) # Add the number to our list
        
    elif choice == '2':
        # Header for the table
        print(f"{'Original'} | {'Discount'} | {'Subtotal'}")
        
        
        total_revenue = 0 # Reset revenue before re-calculating the list
        
        for order_amount in orders: # Use 'amt' to avoid confusing Python
            if order_amount >= 2000:
                discount = order_amount * 15/100
                sub_total = order_amount - discount
            elif order_amount>=1500 and order_amount < 2000:
                discount = order_amount * 10/100
                sub_total = order_amount - discount
            elif  order_amount>=1000 and order_amount < 1500:
                discount = order_amount * 7/100
                sub_total = order_amount - discount
            elif order_amount < 1000:
                discount = 0
                sub_total = order_amount - discount
            
            #sub_total = order_amount - discount
            total_revenue += sub_total
            
            # Print the summary row
            print(f"{order_amount} | {discount} | {sub_total}")
            
        
        print(f"Total Revenue: {total_revenue}")
```
